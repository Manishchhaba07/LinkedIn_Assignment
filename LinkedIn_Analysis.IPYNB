# 👥 LinkedIn Adjacency List Generator (Undirected)
This script builds a complete **undirected adjacency list** from cleaned `.csv` files stored in the `Cleaned_LinkedIn_CSVs` folder.



```python
import os
import pandas as pd
import json
from collections import defaultdict

# 📁 Set your folder path
folder_path = "Cleaned_LinkedIn_CSVs"

# ✅ Use defaultdict to automatically create empty lists
adjacency_list = defaultdict(set)

# 🔁 Read each student's CSV
for file_name in os.listdir(folder_path):
    if file_name.endswith(".csv") and not file_name.startswith("all_students"):
        student_name = file_name.replace(".csv", "").strip().title()

        try:
            df = pd.read_csv(os.path.join(folder_path, file_name))
            connections = (df["First Name"].astype(str) + " " + df["Last Name"].astype(str)).str.title().tolist()

            # Add student → connections
            for connection in connections:
                if connection != student_name:
                    adjacency_list[student_name].add(connection)
                    adjacency_list[connection].add(student_name)  # Ensure it's undirected

        except Exception as e:
            print(f"⚠️ Error in file {file_name}: {e}")

# ✅ Convert sets to lists for JSON compatibility
final_adj_list = {k: sorted(list(v)) for k, v in adjacency_list.items()}

# 💾 Save to JSON
output_file = os.path.join(folder_path, "complete_adjacency_list.json")
with open(output_file, "w", encoding="utf-8") as f:
    json.dump(final_adj_list, f, indent=4)

# ✅ Display full output in terminal
print(f"\n✅ Adjacency list created with {len(final_adj_list)} students.")
print(f"📁 Saved to: {output_file}\n")
print("🔍 Full adjacency list:")

for student, connections in final_adj_list.items():
    print(f"{student} → {connections}")
```

# 📊 **Connection Degrees Calculator for LinkedIn CSV Files**
This script processes `.csv` files stored in a specified folder, calculates the **connection degree** (number of non-empty rows) for each student, and outputs the results in a **sorted JSON format**.




```python
import os
import csv
import json

def get_connection_degrees(folder_path, output_json='all_students_degree.json'):
    degree_dict = {}

    for filename in os.listdir(folder_path):
        if filename.endswith(".csv"):
            file_path = os.path.join(folder_path, filename)
            try:
                with open(file_path, mode='r', encoding='utf-8') as file:
                    reader = csv.reader(file)
                    header = next(reader, None)
                    degree = sum(1 for _ in reader)
                    degree_dict[filename] = degree
            except UnicodeDecodeError:
                # Fallback if UTF-8 fails
                with open(file_path, mode='r', encoding='ISO-8859-1') as file:
                    reader = csv.reader(file)
                    header = next(reader, None)
                    degree = sum(1 for _ in reader)
                    degree_dict[filename] = degree

    # Sort dictionary by degree value in descending order
    sorted_degrees = dict(sorted(degree_dict.items(), key=lambda item: item[1], reverse=True))

    # Save to JSON
    json_path = os.path.join(folder_path, output_json)
    with open(json_path, 'w', encoding='utf-8') as f:
        json.dump(sorted_degrees, f, indent=4)

    print(f"[✔] Saved degree list to: {json_path}")
    return sorted_degrees

# Example usage:
folder = "Cleaned_LinkedIn_CSVs"
get_connection_degrees(folder)
```

# 🔍 LinkedIn Graph Path Explorer
This script loads an **adjacency list** of LinkedIn connections from a JSON file and explores paths between two students using:



```python
import json
import random
from collections import deque

# Load adjacency list from JSON file
json_path = r"D:\MFC ASSIGNMENT\LinkedIn Data Public\adjacency_list.json"

try:
    with open(json_path, 'r', encoding='utf-8') as f:
        graph = json.load(f)
except Exception as e:
    print(f"❌ Error loading JSON: {e}")
    exit()

# Smart random walk that ensures it ends at the destination
def random_walk(graph, start, end, max_steps=100):
    path = [start]
    current = start

    for _ in range(max_steps):
        neighbors = graph.get(current, [])
        if not neighbors:
            break

        # if end in neighbors:
        #     path.append(end)
        #     return path

        next_node = random.choice(neighbors)
        path.append(next_node)
        current = next_node

    # Final check: if end is not already in path, append it
    if path[-1] != end:
        path.append(end)

    return path

# Remove cycles from path
def prune_path(path):
    visited = {}
    pruned = []

    for node in path:
        if node in visited:
            cycle_start = visited[node]
            pruned = pruned[:cycle_start + 1]
        else:
            visited[node] = len(pruned)
            pruned.append(node)

    # Ensure the end node is the last one
    if pruned[-1] != path[-1]:
        pruned.append(path[-1])

    return pruned

# Find the shortest path using BFS
def bfs_shortest_path(graph, start, end):
    queue = deque([[start]])
    visited = set()

    while queue:
        path = queue.popleft()
        node = path[-1]

        if node == end:
            return path

        if node not in visited:
            visited.add(node)
            for neighbor in graph.get(node, []):
                if neighbor not in visited:
                    new_path = list(path)
                    new_path.append(neighbor)
                    queue.append(new_path)

    return None

# Inputs
start_student = "Shimaila Praveen"
end_student = "Dhairya Seth"

# Run 3 random walks
random_walks = []
for i in range(3):
    walk = random_walk(graph, start_student, end_student)
    pruned = prune_path(walk)
    random_walks.append({
        "original": walk,
        "pruned": pruned,
        "length": len(pruned),
    })

# Find shortest pruned path
shortest_pruned = min(random_walks, key=lambda x: x["length"])

# Find actual shortest path
true_shortest = bfs_shortest_path(graph, start_student, end_student)

# Output
for idx, walk in enumerate(random_walks, start=1):
    print(f"\n🔁 Random Walk {idx}:")
    print(" -> ".join(walk["original"]))
    print(f"🧹 Pruned: {' -> '.join(walk['pruned'])}")
    print(f"🔢 Pruned Length: {walk['length']}")

print("\n✅ Shortest of the 3 Pruned Random Walks:")
print(" -> ".join(shortest_pruned["pruned"]))
print(f"Length: {shortest_pruned['length']}")

if true_shortest:
    print("\n📏 True Shortest Path (BFS):")
    print(" -> ".join(true_shortest))
    print(f"Length: {len(true_shortest)}")
else:
    print("\n❌ No path found using BFS.")
```

# 📈 LinkedIn Graph Overview
This script analyzes the **complete adjacency list** of LinkedIn connections and provides a quick summary of the graph's size and density.



```python
import json

# Load the adjacency list safely using raw string
with open(r"LinkedIn Data Public\complete_adjacency_list.json", "r", encoding="utf-8") as f:
    adjacency_list = json.load(f)

# Total unique students (nodes)
total_students = len(adjacency_list)

# Total connections (sum of all edges)
total_connections = sum(len(conns) for conns in adjacency_list.values())

# Average connections per student
average_connections = total_connections / total_students if total_students else 0

print(f"Total unique students: {total_students}")
print(f"Total number of connections: {total_connections}")
print(f"Average connections per student: {average_connections:.2f}")
```
